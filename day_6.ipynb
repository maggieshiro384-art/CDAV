```python
list_val_neg = []
list_val_pos=[]
list_val_zer=[]
while True:
    num =(input("Enter a number or = to break: "))


    #stopping condition
    if num == "=":
        break
    val=int(num)
    if val <0:
      list_val_neg.append(val)
    elif val==0:
      list_val_zer.append(val)
    else:
      list_val_pos.append(val)



for n in list_val_neg:
  print(n)
for z in list_val_zer:
  print(z)
for p in list_val_pos:
  print(p)





```
