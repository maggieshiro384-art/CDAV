# 1. Plot survival rate by passenger class
*   Dataset = titanic
*   Barplot to compare categories(the three classes)





```python
titanic.columns
```

```python
sns.barplot(data=titanic, x="class", y="survived", palette="pastel")
```

#### 1.  The barplot calculates the rate on its own using the dat acolected from the survived column

#### 2.  The survival rate decreases from first to third class

# 2. Compare the survival rate by sex within each class

```python
fig, ax = plt.subplots(figsize=(10,6))

sns.barplot(data=titanic, x="class", y="survived", hue="sex", palette="pastel")
```

#### Females have a higher survival rate than males across all three classes

#### The error bars are relatively short suggesting that the survival rate is not as variable

# 3. Show fare vs age with a regression line; comment on the trend and any outliers

```python
sns.regplot(data=titanic, x="age", y="fare")
plt.show()
```

#### There appears to be a slight positive relationship between age and fare:

#### As age increases, fare tends to increase slightly.

#### However, the points are very spread out, so the relationship isn't particularly strong.

#### You can also see a few very high-fare outliers, especially around ages 30–40 where fares go above 500.
