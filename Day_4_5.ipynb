```python
message = input("Enter your message: ")
shift = int(input("Enter shift value (e.g., 3 or -3): "))

result = ""

for ch in message:
    if ch.isalpha():
        if ch.islower():
            base = ord('a')
        else:
            base = ord('A')

        new_char = chr((ord(ch) - base + shift) % 26 + base)
        result += new_char
    else:

        result += ch

print("Shifted message:", result)
```

```python
m = int(input("Enter first number: "))
n = int(input("Enter second number: "))

# Initialize d
if m < n:
    d = m
else:
    d = n

not_gcd = True   # assume we haven't found GCD yet

while not_gcd:
    if m % d == 0 and n % d == 0:
        print("GCD is:", d)
        not_gcd = False   # stop the loop
    else:
        d = d - 1
```

The for loop

The single argument range
Start from zero by default and 5 is not printed(its exclusive)

```python
for number in range(5):
  print(number)
```

The two argument range
it shows where to start and where to stop. 6 is not printed its exclusive
range(1,6)

```python
for number in range (1,6):
  print(number)
```

The three argument first number shows where to start, second shows where it will end and its exclusive and the third number shows the increament value

```python
for number in range (10,0,-1):
  print(number)
```
